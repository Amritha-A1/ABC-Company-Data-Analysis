```python
df['College']=df['College'].fillna(df['College'].mode()[0])
df['Salary']=df['Salary'].fillna(df['Salary'].mean())
df.isna().sum()
```

```python
#Correct the data in the "height" column by replacing it with random numbers between 150 and 180.
df['Height']=np.random.randint(150,181,size=len(df))
df
```

```python
#Determine the distribution of employees across each team and calculate the percentage split relative to the total number of employees.
team_count=df['Team'].value_counts()
percentage_split=team_count/len(df)*100
distribution =pd.DataFrame({"Employee Count": team_count,"Percentage":percentage_split})
print(distribution)


plt.figure(figsize=(10, 6))
sns.barplot(x="Team", y="Percentage", data=distribution)
plt.title('Percentage Distribution of Employees Across Teams')
plt.xlabel('Team')
plt.ylabel('Percentage')
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x="Team", y="Employee Count", data=distribution)
plt.title('Employee Distribution by Teams')
plt.xlabel('Team')
plt.ylabel('Number of Employess')
plt.xticks(rotation=90)
plt.show()
```

```python
#Segregate employees based on their positions within the company.
position_count=df['Position'].value_counts().reset_index()
print(position_count)
plt.figure(figsize=(10, 6))
sns.barplot(x="Position", y="count", data=position_count)
plt.title('Distribution of Employees Across Positions')
plt.xlabel('Position')
plt.ylabel('Employee Count')
#grid=plt.grid(True,alpha=0.3)
plt.show()
```

```python
#Identify the predominant age group among employees.
age_groups = pd.cut(
    df['Age'],
    bins=[18, 25, 32, 40],
    labels=['18-25', '26-32', '33-40']
)

age_group_count = age_groups.value_counts().sort_index()
print(age_group_count)
print("\nPredominant Age Group:", age_group_count.idxmax())
plt.figure(figsize=(10, 6))
sns.barplot(x=age_group_count.index, y=age_group_count.values)
plt.show()
```

```python
# Discover which team and position have the highest salary expenditure.

team_salary=df.groupby('Team')['Salary'].sum().reset_index()
print("Salary expenditure by team:")
print(team_salary)
print("\n")
highest_salary_team=team_salary.loc[team_salary['Salary'].idxmax()]
print("Team with highest salary expenditure:")
print(highest_salary_team)
print("\n")
plt.figsize=(10,6)
sns.barplot(x='Team',y='Salary',data=team_salary)
plt.title('Salary Expenditure by Team')
plt.xlabel('Team')
plt.ylabel('Total_Salary')
plt.xticks(rotation=90)
plt.show()
```

```python
position_salary=df.groupby('Position')['Salary'].sum().reset_index()
print("Salary expenditure by position:")
print(position_salary)
print("\n")

highest_salary_position=position_salary.loc[position_salary['Salary'].idxmax()]
print("Position with highest salary expenditure:")
print(highest_salary_position)
plt.figsize=(10,6)
sns.barplot(x='Position',y='Salary',data=position_salary)
plt.title('Salary Expenditure by Position')
plt.xlabel('Position')
plt.ylabel('Total_Salary')
plt.show()
```
